## 2. Hoàn thiện starter

Mở `bai-tap/buoi4_perceptron_cit.py`, hoàn thiện TODO 1–3, lưu file rồi chạy lại hai cell bên dưới.

```python
from importlib.util import module_from_spec, spec_from_file_location
from pathlib import Path

candidates = [
    Path('bai-tap/buoi4_perceptron_cit.py'),
    Path('../bai-tap/buoi4_perceptron_cit.py'),
]
starter_path = next((path for path in candidates if path.exists()), None)
assert starter_path is not None, 'Không tìm thấy starter. Hãy mở notebook từ repo lac-hong-neural-network.'

spec = spec_from_file_location('buoi4_perceptron_cit', starter_path)
starter = module_from_spec(spec)
spec.loader.exec_module(starter)
print(f'Đã nạp: {starter_path}')
```

```python
starter.run_checks()
print('✓ Ba TODO đã vượt qua kiểm tra.')
```

## 3. Đối chiếu một epoch với bảng tính tay

```python
w_final, b_final, history = starter.train_one_epoch()
for row in history:
    print(row)
print('w cuối =', w_final, '| b cuối =', b_final)
```

**Giải thích bằng lời:** lượt nào làm trọng số thay đổi? Feature đang bật đã kéo trọng số theo hướng nào?

> Viết câu trả lời 3–5 dòng tại đây.

## 4. Thí nghiệm learning rate

Chạy cùng một epoch với ba giá trị $\eta$. Quan sát độ lớn bước cập nhật; không kết luận tốc độ hội tụ chỉ từ một epoch.

```python
for eta in (0.1, 0.5, 1.0):
    w_eta, b_eta, rows_eta = starter.train_one_epoch(eta=eta)
    errors = [row['error'] for row in rows_eta]
    print(f'eta={eta}: errors={errors}, w={w_eta}, b={b_eta:.2f}')
```

**Kết luận:** khi $\eta$ tăng, độ lớn mỗi bước cập nhật thay đổi thế nào?

> Viết câu trả lời 2–3 dòng tại đây.
